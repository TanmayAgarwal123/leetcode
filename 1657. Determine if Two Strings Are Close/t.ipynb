```python
class Solution:
    def closeStrings(self, word1: str, word2: str) -> bool:
        if len(word1)!=len(word2):
            return False
        else:
            v1=Counter(word1)
            v2=Counter(word2)
            if sorted(v1.keys()) == sorted(v2.keys()) and sorted(v1.values()) == sorted(v2.values()):
                return True
        return False
```
