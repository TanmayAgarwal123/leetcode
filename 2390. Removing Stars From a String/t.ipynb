```python
class Solution:
    def removeStars(self, s: str) -> str:
        ans=[]
        for i in s:
            if i is '*':
                ans.pop()
            else:
                ans+=i
        return ''.join(ans)


    #this one takes more time, cause we are adding two lists instead of a element to a list
```

```python
class Solution:
    def removeStars(self, s: str) -> str:
        ans=[]
        for i in s:
            if i is '*':
                ans.pop()
            else:
                ans.append(i)
        return ''.join(ans)
```
